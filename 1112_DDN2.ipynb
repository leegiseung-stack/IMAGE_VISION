```python
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten
from keras.utils import to_categorical
import numpy as np
import matplotlib.pyplot as plt
```

```python
(X_train,y_train),(X_test,y_test)=mnist.load_data()
```

```python
X_train=X_train.reshape(60000,784).astype("float32")
X_test=X_test.reshape(10000,784).astype("float32")
```

```python
X_train[0]
```

```python
X_train /= 255
X_test  /= 255
#민맥스스케일러
```

```python
y_train = to_categorical(y_train, 10)
y_test  = to_categorical(y_test, 10)
```

```python
y_train[0]
```

```python
y_train.shape
```

```python
y_train[0:10]
```

```python
#원핫인코딩
#사이킷런 원핫인코딩
#판다스 겟더미스
#케라스 투 카테고리컬 
```

```python
from keras.layers import Activation

model = Sequential()
model.add(Dense(512, input_shape=(784,)))
model.add(Activation("relu"))
model.add(Dropout(0.2))
model.add(Dense(512))
model.add(Activation("relu"))
model.add(Dropout(0.2))
model.add(Dense(10))
model.add(Activation("softmax"))
model.summary()
```

```python
#모델 설계 -> 컴파일 -> 학습 
```

```python
model.compile(loss="categorical_crossentropy",
              optimizer="adam",
              metrics=["accuracy"])
```

```python
hist = model.fit(X_train, y_train, epochs=5,batch_size=32)
```

```python
hist.history
#평가결과는 튜플형태로나옴 
```

```python
score=model.evaluate(X_test,y_test,verbose=1)
print("\n loss=",score[0] )
print("accuracy",score[1])
```

```python
model.save("mnist_keras.keras")
```

```python
###keros 버전 BMI추가예제

from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten
from keras.callbacks import EarlyStopping
from keras.utils import to_categorical
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
csv=pd.read_csv("bmi.csv")
csv.head(2)
```

```python
from sklearn.preprocessing import MinMaxScaler

mm=MinMaxScaler()
csv["weight"] = mm.fit_transform(csv[["weight"]])
csv["height"] = mm.fit_transform(csv[["height"]])

```

```python
from sklearn.preprocessing import LabelEncoder

encoder=LabelEncoder()
csv["label"] = encoder.fit_transform(csv["label"])
csv
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(csv.iloc[:, :-1], csv.iloc[:, -1], random_state=42)
X_train.shape, X_test.shape
```

```python
y_train.shape,y_test.shape
```

```python
#원핫인코딩
y_train=to_categorical(y_train,3)
y_test=to_categorical(y_test,3)
#3가지클래스로 원핫인코딩 
```

```python
from keras.layers import Activation

model = Sequential()
model.add(Dense(512, activation="relu", input_shape=(2,)))
#입력층
model.add(Dropout(0.1))
model.add(Dense(512,activation="relu"))
#첫번째은닉층
model.add(Dropout(0.1))
#두번째은닉층
model.add(Dense(3,activation="softmax"))
#출력층
model.summary()
```

```python
model.compile(loss="categorical_crossentropy",
              optimizer="adam",
              metrics=["accuracy"])
```

```python

hist = model.fit(X_train, y_train,
                 batch_size=100,
                 epochs=20,)
#만오천개의 데이터에서 100개씩학습해서 150
```

```python
score = model.evaluate(X_test, y_test, verbose=1)
print("\nloss =", score[0])
print("accuracy =", score[1])
```

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
```

```python
fashion_mnist = keras.datasets.fashion_mnist
(train_image, train_label), (test_image, test_label) = fashion_mnist.load_data()
```

```python
class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
```

```python
train_image.shape,test_image.shape
```

```python
train_label.shape,test_label.shape
```

```python
train_label[:10]
```

```python
plt.figure()
plt.imshow(train_image[0])
plt.colorbar()
plt.grid(False)
plt.show()
```

```python
train_image = train_image / 255.0
test_image  = test_image / 255.0
#민맥스스케일링효과
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,10))
for index in range(25):
    plt.subplot(5,5,index+1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_image[index], cmap=plt.cm.binary)
    #바이너리가 이진이라 흑백
    plt.xlabel(class_names[train_label[index]])
plt.show()
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28,28)),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])
```

```python
model.compile(optimizer="adam",
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])
```

```python
history=model.fit(train_image, train_label, epochs=5)
```

```python
plt.plot(history.history["loss"])
plt.plot(history.history["accuracy"])
plt.legend(["loss","accuracy"])
plt.show()
```

```python
test_loss,test_acc=model.evaluate(test_image,test_label,verbose=2)
print("\n 테스트 정확도",test_acc)
```
