```python
#상수노드
import tensorflow as tf
tf.compat.v1.disable_eager_execution()

a=tf.constant(1.0)
b=tf.constant(2.0)
c=tf.constant([[1.0,2.0],[3.0,4.0]])
# = 연산대입이아니고 연산정의임
print(a)
print(a+b)
print(c)

sess=tf.compat.v1.Session()
print(sess.run([a,b]))
print(sess.run(c))
print(sess.run([a+b]))
print(sess.run(c+1.0))

sess.close()
```
